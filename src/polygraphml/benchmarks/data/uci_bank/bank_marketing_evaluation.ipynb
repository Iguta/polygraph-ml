# UCI Bank Marketing evaluation

This notebook evaluates whether a client subscribes to a term deposit. The model-ready evaluation table and category mapping are reproducibly derived by `scripts/generate_uci_bank_benchmark.py`.

```
import pandas as pd
import skops.io as sio
from sklearn.metrics import roc_auc_score
df = pd.read_csv('bank_marketing_evaluation.csv')
model = sio.load('bank_marketing_model.skops', trusted=[] )
```

```
features = ['age', 'job', 'marital', 'education', 'default', 'balance', 'housing', 'loan', 'contact', 'day', 'month', 'duration', 'campaign', 'pdays', 'previous', 'poutcome']
test = df['split'] == 'test'
probabilities = model.predict_proba(df.loc[test, features])[:, 1]
```

```
auc = roc_auc_score(df.loc[test, 'subscribed'], probabilities)
print(f'ROC AUC: {auc:.6f}')
```
